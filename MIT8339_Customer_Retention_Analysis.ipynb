```python

# ==========================================
# MIT 8339 - CUSTOMER RETENTION PROJECT
# STEP 1: DATASET INSPECTION
# ==========================================

import pandas as pd
import numpy as np

# Load the original dataset
df = pd.read_csv('/content/BankChurners.csv')

# Display dataset dimensions
print("DATASET OVERVIEW")
print("================")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

# Display the first five records
print("\nFIRST FIVE RECORDS")
display(df.head())

# Display column names and data types
print("\nCOLUMN INFORMATION")
df.info()

# Check for missing values
print("\nMISSING VALUES")
display(df.isnull().sum())

# Check for duplicate records
print("\nDUPLICATE RECORDS")
print(df.duplicated().sum())

# Display basic statistical information
print("\nSUMMARY STATISTICS")
display(df.describe(include='all').T)

print("\nINITIAL DATA INSPECTION COMPLETED!")
```

```python

# ==========================================
# MIT 8339 - DATA QUALITY ASSESSMENT
# STEP 2
# ==========================================

# 1. Check all missing values
print("MISSING VALUE SUMMARY")
print(df.isnull().sum().to_string())

# 2. Check duplicate records
print("\nDUPLICATE RECORD CHECK")
print("Total duplicate rows:", df.duplicated().sum())

# 3. Check unique values in important categories
print("\nATTRITION CATEGORIES")
print(df['Attrition_Flag'].value_counts())

print("\nGENDER CATEGORIES")
print(df['Gender'].value_counts())

print("\nCARD CATEGORIES")
print(df['Card_Category'].value_counts())

print("\nINCOME CATEGORIES")
print(df['Income_Category'].value_counts())

# 4. Check numerical data ranges
print("\nNUMERICAL DATA SUMMARY")
display(df.describe().T)

# 5. Check all column names
print("\nALL DATASET COLUMNS")
for column in df.columns:
    print(column)

print("\nDATA QUALITY ASSESSMENT COMPLETED!")
```

```python

# ==========================================
# MIT 8339 - DETAILED DATA QUALITY CHECK
# STEP 3
# ==========================================

# 1. Check whether customer IDs are unique
print("CUSTOMER ID CHECK")
print("Unique customer IDs:", df['CLIENTNUM'].nunique())
print("Total records:", len(df))

# 2. Investigate Unknown categories
print("\nUNKNOWN CATEGORY CHECK")

for column in df.select_dtypes(include='object').columns:
    unknown_count = df[column].astype(str).str.strip().str.lower().eq('unknown').sum()
    if unknown_count > 0:
        print(column, ":", unknown_count, "Unknown records")

# 3. Check numerical minimum and maximum values
print("\nNUMERICAL RANGE CHECK")

numeric_columns = df.select_dtypes(include='number').columns

display(df[numeric_columns].agg(['min', 'max']).T)

# 4. Check data types
print("\nDATA TYPES")
display(df.dtypes)

# 5. Inspect the two classifier columns
print("\nCLASSIFIER COLUMN CHECK")

classifier_columns = [
    col for col in df.columns
    if 'Naive_Bayes_Classifier' in col
]

print("Classifier columns found:", len(classifier_columns))
for col in classifier_columns:
    print(col)

print("\nDETAILED INSPECTION COMPLETED!")
```

```python

# =================================================
# MIT 8339 - CUSTOMER RETENTION PROJECT
# STAGE 3: DATA CLEANING
# =================================================

import pandas as pd
import numpy as np

# 1. Create a separate working copy
clean_df = df.copy()

print("STARTING DATA CLEANING")
print("Original rows:", len(clean_df))
print("Original columns:", len(clean_df.columns))

# 2. Remove pre-existing classifier columns
classifier_columns = [
    col for col in clean_df.columns
    if 'Naive_Bayes_Classifier' in col
]

clean_df = clean_df.drop(columns=classifier_columns)

print("\nClassifier columns removed:", len(classifier_columns))

# 3. Remove accidental spaces from column names
clean_df.columns = clean_df.columns.str.strip()

# 4. Standardise text values
text_columns = clean_df.select_dtypes(
    include=['object']
).columns

for col in text_columns:
    clean_df[col] = clean_df[col].str.strip()

# 5. Check missing values again
print("\nMISSING VALUES AFTER CLEANING")
print(clean_df.isnull().sum().sum())

# 6. Check duplicate records again
print("\nDUPLICATES AFTER CLEANING")
print(clean_df.duplicated().sum())

# 7. Verify customer ID uniqueness
print("\nUNIQUE CUSTOMER IDs")
print(clean_df['CLIENTNUM'].nunique())

# 8. Display cleaned dataset structure
print("\nCLEANED DATASET OVERVIEW")
print("Rows:", clean_df.shape[0])
print("Columns:", clean_df.shape[1])

display(clean_df.head())

# 9. Export cleaned dataset
clean_df.to_csv(
    '/content/BankChurners_Cleaned.csv',
    index=False
)

print("\nCLEANING COMPLETED SUCCESSFULLY!")
print("Cleaned dataset exported.")
```

```python

# =====================================================
# MIT 8339 - CUSTOMER RETENTION PROJECT
# STAGE 4: CUSTOMER SEGMENTATION
# =====================================================

# Work on another copy of the cleaned dataset
analysis_df = clean_df.copy()

# 1. AGE BAND
analysis_df['Age_Band'] = pd.cut(
    analysis_df['Customer_Age'],
    bins=[0, 34, 54, 100],
    labels=['Young Adult', 'Middle Aged', 'Older Adult']
)

# 2. INCOME SEGMENT
income_mapping = {
    'Less than $40K': 'Low Income',
    '$40K - $60K': 'Middle Income',
    '$60K - $80K': 'Middle Income',
    '$80K - $120K': 'High Income',
    '$120K +': 'High Income',
    'Unknown': 'Unknown'
}

analysis_df['Income_Segment'] = (
    analysis_df['Income_Category']
    .map(income_mapping)
    .fillna('Unknown')
)

# 3. CREDIT UTILISATION CATEGORY
analysis_df['Utilisation_Category'] = pd.cut(
    analysis_df['Avg_Utilization_Ratio'],
    bins=[-0.001, 0.30, 0.70, 1.001],
    labels=['Low Utilisation', 'Moderate Utilisation', 'High Utilisation']
)

# 4. TRANSACTION ACTIVITY CATEGORY
activity_percentile = analysis_df['Total_Trans_Ct'].rank(
    pct=True, method='average'
)

analysis_df['Transaction_Activity'] = np.select(
    [
        activity_percentile <= 0.25,
        activity_percentile <= 0.75
    ],
    [
        'Low Activity',
        'Moderate Activity'
    ],
    default='High Activity'
)

# 5. CUSTOMER VALUE CLASS
# Transaction amount is used as a proxy for customer value.
value_percentile = analysis_df['Total_Trans_Amt'].rank(
    pct=True, method='average'
)

analysis_df['Customer_Value_Class'] = np.select(
    [
        value_percentile <= 0.33,
        value_percentile <= 0.67
    ],
    [
        'Low Value',
        'Medium Value'
    ],
    default='High Value'
)

# 6. RULE-BASED RETENTION RISK SCORE
# Each behavioural indicator contributes one point.

analysis_df['Risk_Score'] = (
    (analysis_df['Months_Inactive_12_mon'] >= 3).astype(int)
    + (analysis_df['Contacts_Count_12_mon'] >= 3).astype(int)
    + (analysis_df['Total_Ct_Chng_Q4_Q1'] < 0.5).astype(int)
    + (analysis_df['Total_Trans_Ct'] < analysis_df['Total_Trans_Ct'].median()).astype(int)
)

analysis_df['Retention_Risk_Band'] = np.select(
    [
        analysis_df['Risk_Score'] >= 3,
        analysis_df['Risk_Score'] == 2
    ],
    [
        'High Risk',
        'Medium Risk'
    ],
    default='Low Risk'
)

# DISPLAY THE NEW FIELDS
print("CUSTOMER SEGMENTATION COMPLETED")
print("================================")

new_columns = [
    'Age_Band',
    'Income_Segment',
    'Utilisation_Category',
    'Transaction_Activity',
    'Customer_Value_Class',
    'Risk_Score',
    'Retention_Risk_Band'
]

display(analysis_df[new_columns].head(10))

# CUSTOMER DISTRIBUTIONS
for column in new_columns:
    print("\n", column)
    print(analysis_df[column].value_counts().to_string())

print("\nSEGMENTATION COMPLETED SUCCESSFULLY!")
```

```python

# =====================================================
# MIT 8339 - CUSTOMER RETENTION PROJECT
# STAGE 5: KEY PERFORMANCE INDICATORS (KPIs)
# =====================================================

# 1. Overall customer statistics

total_customers = len(analysis_df)

attrited = (
    analysis_df['Attrition_Flag'] == 'Attrited Customer'
).sum()

existing = (
    analysis_df['Attrition_Flag'] == 'Existing Customer'
).sum()

attrition_rate = (attrited / total_customers) * 100
retention_rate = (existing / total_customers) * 100

# 2. Calculate average banking indicators

average_age = analysis_df['Customer_Age'].mean()
average_credit_limit = analysis_df['Credit_Limit'].mean()
average_utilisation = analysis_df['Avg_Utilization_Ratio'].mean()
average_transaction_amount = analysis_df['Total_Trans_Amt'].mean()
average_transaction_count = analysis_df['Total_Trans_Ct'].mean()
average_relationships = analysis_df['Total_Relationship_Count'].mean()

# 3. Display KPI results

print("=" * 50)
print("BANK CUSTOMER RETENTION KPI REPORT")
print("=" * 50)

print(f"Total Customers: {total_customers:,}")
print(f"Existing Customers: {existing:,}")
print(f"Attrited Customers: {attrited:,}")

print(f"\nCustomer Attrition Rate: {attrition_rate:.2f}%")
print(f"Customer Retention Rate: {retention_rate:.2f}%")

print("\nAVERAGE CUSTOMER INDICATORS")
print("-" * 50)

print(f"Average Customer Age: {average_age:.2f}")
print(f"Average Credit Limit: ${average_credit_limit:,.2f}")
print(f"Average Credit Utilisation: {average_utilisation:.2%}")
print(f"Average Transaction Amount: ${average_transaction_amount:,.2f}")
print(f"Average Transaction Count: {average_transaction_count:.2f}")
print(f"Average Product Relationships: {average_relationships:.2f}")

# 4. Compare existing and attrited customers

print("\nCUSTOMER GROUP COMPARISON")
print("-" * 50)

comparison = analysis_df.groupby(
    'Attrition_Flag'
)[[
    'Customer_Age',
    'Credit_Limit',
    'Avg_Utilization_Ratio',
    'Total_Trans_Amt',
    'Total_Trans_Ct',
    'Months_Inactive_12_mon',
    'Contacts_Count_12_mon',
    'Total_Relationship_Count'
]].mean().round(2)

display(comparison)

print("\nKPI CALCULATION COMPLETED!")
```

```python

# =====================================================
# MIT 8339 - CUSTOMER RETENTION PROJECT
# STAGE 5.2: DEMOGRAPHIC ATTRITION ANALYSIS
# =====================================================

# Define the categories we want to investigate

categories = [
    'Age_Band',
    'Gender',
    'Income_Category',
    'Card_Category'
]

# Calculate customer counts and attrition rates

for column in categories:

    print("\n" + "=" * 55)
    print("ATTRITION ANALYSIS BY", column.upper())
    print("=" * 55)

    # Count existing and attrited customers
    counts = pd.crosstab(
        analysis_df[column],
        analysis_df['Attrition_Flag']
    )

    # Calculate attrition rate within each category
    rates = pd.crosstab(
        analysis_df[column],
        analysis_df['Attrition_Flag'],
        normalize='index'
    ) * 100

    # Combine the results
    results = pd.DataFrame({
        'Total_Customers': counts.sum(axis=1),
        'Attrited_Customers': counts['Attrited Customer'],
        'Existing_Customers': counts['Existing Customer'],
        'Attrition_Rate_Percent': rates['Attrited Customer']
    })

    display(results.round(2))

# Save the demographic analysis results

print("\nDEMOGRAPHIC ANALYSIS COMPLETED!")
```

```python

# =====================================================
# MIT 8339 - CUSTOMER RETENTION PROJECT
# STAGE 5.3: BEHAVIOURAL ATTRITION ANALYSIS
# =====================================================

# 1. Create inactivity groups
analysis_df['Inactivity_Group'] = pd.cut(
    analysis_df['Months_Inactive_12_mon'],
    bins=[-1, 1, 2, 3, 6],
    labels=['0-1 Months', '2 Months', '3 Months', '4-6 Months']
)

# 2. Create contact frequency groups
analysis_df['Contact_Group'] = pd.cut(
    analysis_df['Contacts_Count_12_mon'],
    bins=[-1, 1, 2, 3, 6],
    labels=['0-1 Contacts', '2 Contacts', '3 Contacts', '4-6 Contacts']
)

# 3. Create transaction frequency groups
analysis_df['Transaction_Group'] = pd.cut(
    analysis_df['Total_Trans_Ct'],
    bins=[0, 39, 59, 79, 200],
    labels=['Low (10-39)', 'Moderate (40-59)',
            'High (60-79)', 'Very High (80+)']
)

# 4. Analyse attrition across the three behaviours

behaviour_columns = [
    'Inactivity_Group',
    'Contact_Group',
    'Transaction_Group'
]

for column in behaviour_columns:

    print("\n" + "=" * 55)
    print("ATTRITION BY", column.upper())
    print("=" * 55)

    counts = pd.crosstab(
        analysis_df[column],
        analysis_df['Attrition_Flag']
    )

    rates = pd.crosstab(
        analysis_df[column],
        analysis_df['Attrition_Flag'],
        normalize='index'
    ) * 100

    results = pd.DataFrame({
        'Total_Customers': counts.sum(axis=1),
        'Attrited_Customers': counts['Attrited Customer'],
        'Existing_Customers': counts['Existing Customer'],
        'Attrition_Rate_Percent': rates['Attrited Customer']
    })

    display(results.round(2))

# 5. Compare average behavioural indicators

print("\nAVERAGE BEHAVIOURAL INDICATORS BY CUSTOMER STATUS")

behaviour_comparison = analysis_df.groupby(
    'Attrition_Flag'
)[[
    'Months_Inactive_12_mon',
    'Contacts_Count_12_mon',
    'Total_Trans_Ct',
    'Total_Ct_Chng_Q4_Q1',
    'Total_Amt_Chng_Q4_Q1'
]].mean().round(3)

display(behaviour_comparison)

print("\nBEHAVIOURAL ANALYSIS COMPLETED!")
```

```python

import ipywidgets as widgets
from IPython.display import display, clear_output, HTML
import plotly.express as px
import pandas as pd

# -----------------------------
# 1. CREATE INTERACTIVE FILTERS
# -----------------------------

gender_filter = widgets.Dropdown(
    options=['All'] + sorted(df['Gender'].dropna().unique().tolist()),
    value='All',
    description='Gender:'
)

income_filter = widgets.Dropdown(
    options=['All'] + sorted(df['Income_Category'].dropna().unique().tolist()),
    value='All',
    description='Income:'
)

card_filter = widgets.Dropdown(
    options=['All'] + sorted(df['Card_Category'].dropna().unique().tolist()),
    value='All',
    description='Card:'
)

filters = widgets.HBox([
    gender_filter,
    income_filter,
    card_filter
])

# -----------------------------
# 2. CREATE DASHBOARD OUTPUTS
# -----------------------------

executive_output = widgets.Output()
retention_output = widgets.Output()

tabs = widgets.Tab(children=[
    executive_output,
    retention_output
])

tabs.set_title(0, 'Executive Overview')
tabs.set_title(1, 'Retention Analysis')

# -----------------------------
# 3. FILTER DATA
# -----------------------------

def get_filtered_data():
    filtered = df.copy()

    if gender_filter.value != 'All':
        filtered = filtered[
            filtered['Gender'] == gender_filter.value
        ]

    if income_filter.value != 'All':
        filtered = filtered[
            filtered['Income_Category'] == income_filter.value
        ]

    if card_filter.value != 'All':
        filtered = filtered[
            filtered['Card_Category'] == card_filter.value
        ]

    return filtered


# -----------------------------
# 4. EXECUTIVE OVERVIEW
# -----------------------------

def show_executive_dashboard(data):

    total = len(data)
    existing = (data['Attrition_Flag'] == 'Existing Customer').sum()
    attrited = (data['Attrition_Flag'] == 'Attrited Customer').sum()

    attrition = (attrited / total * 100) if total else 0
    retention = (existing / total * 100) if total else 0

    avg_age = data['Customer_Age'].mean() if total else 0
    avg_limit = data['Credit_Limit'].mean() if total else 0

    display(HTML(f"""
    <h2>Executive Overview</h2>
    <p>Customer retention and performance summary</p>

    <div style="display:flex; flex-wrap:wrap; gap:12px;">
      <div style="border:1px solid #aaa;padding:14px;border-radius:8px;">
        <b>Total Customers</b><br><h2>{total:,}</h2>
      </div>
      <div style="border:1px solid #aaa;padding:14px;border-radius:8px;">
        <b>Existing Customers</b><br><h2>{existing:,}</h2>
      </div>
      <div style="border:1px solid #aaa;padding:14px;border-radius:8px;">
        <b>Attrited Customers</b><br><h2>{attrited:,}</h2>
      </div>
      <div style="border:1px solid #aaa;padding:14px;border-radius:8px;">
        <b>Attrition Rate</b><br><h2>{attrition:.2f}%</h2>
      </div>
      <div style="border:1px solid #aaa;padding:14px;border-radius:8px;">
        <b>Retention Rate</b><br><h2>{retention:.2f}%</h2>
      </div>
      <div style="border:1px solid #aaa;padding:14px;border-radius:8px;">
        <b>Average Age</b><br><h2>{avg_age:.2f}</h2>
      </div>
      <div style="border:1px solid #aaa;padding:14px;border-radius:8px;">
        <b>Average Credit Limit</b><br><h2>${avg_limit:,.2f}</h2>
      </div>
    </div>
    """))

    if total == 0:
        display(HTML("<p>No customers match the selected filters.</p>"))
        return

    # Customer status pie chart
    status = data['Attrition_Flag'].value_counts().reset_index()
    status.columns = ['Customer Status', 'Customers']

    fig = px.pie(
        status,
        names='Customer Status',
        values='Customers',
        title='Customer Status Distribution',
        hole=0.4
    )
    display(fig)

    # Gender attrition chart
    gender = (
        data.groupby('Gender')['Attrition_Binary']
        .mean().mul(100).reset_index()
    )
    gender.columns = ['Gender', 'Attrition Rate (%)']

    fig = px.bar(
        gender, x='Gender', y='Attrition Rate (%)',
        title='Attrition Rate by Gender',
        text_auto='.2f'
    )
    display(fig)

    # Income attrition chart
    income = (
        data.groupby('Income_Category')['Attrition_Binary']
        .mean().mul(100).reset_index()
    )
    income.columns = ['Income Category', 'Attrition Rate (%)']

    fig = px.bar(
        income, x='Income Category', y='Attrition Rate (%)',
        title='Attrition Rate by Income Category',
        text_auto='.2f'
    )
    display(fig)


# -----------------------------
# 5. RETENTION ANALYSIS
# -----------------------------

def show_retention_dashboard(data):

    display(HTML("""
    <h2>Retention Analysis</h2>
    <p>Customer behaviour, engagement and retention risk</p>
    """))

    if len(data) == 0:
        display(HTML("<p>No customers match the selected filters.</p>"))
        return

    chart_details = [
        ('Inactivity_Group', 'Months of Inactivity',
         'Attrition Rate by Months of Inactivity'),

        ('Contact_Group', 'Customer Service Contacts',
         'Attrition Rate by Customer Service Contacts'),

        ('Transaction_Group', 'Transaction Activity',
         'Attrition Rate by Transaction Activity'),

        ('Retention_Risk_Band', 'Retention Risk Band',
         'Attrition Rate by Retention Risk Band')
    ]

    for column, label, title in chart_details:

        summary = (
            data.groupby(column, observed=False)['Attrition_Binary']
            .mean().mul(100).reset_index()
        )

        summary.columns = [label, 'Attrition Rate (%)']

        fig = px.bar(
            summary,
            x=label,
            y='Attrition Rate (%)',
            title=title,
            text_auto='.2f'
        )

        display(fig)


# -----------------------------
# 6. UPDATE DASHBOARD
# -----------------------------

def update_dashboard(change=None):

    filtered_data = get_filtered_data()

    with executive_output:
        clear_output(wait=True)
        show_executive_dashboard(filtered_data)

    with retention_output:
        clear_output(wait=True)
        show_retention_dashboard(filtered_data)


gender_filter.observe(update_dashboard, names='value')
income_filter.observe(update_dashboard, names='value')
card_filter.observe(update_dashboard, names='value')

# -----------------------------
# 7. DISPLAY DASHBOARD
# -----------------------------

display(HTML("<h1>Nigerian Commercial Bank Customer Retention Dashboard</h1>"))
display(HTML("<p>Use the filters below to explore customer segments.</p>"))
display(filters)
display(tabs)

update_dashboard()
```
